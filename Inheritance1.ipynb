```python
class Number:
    def num(self,x,y):
        self.x = x
        self.y = y
        return f"X : {x} and Y : {y}"
class Sum(Number):
    def add(self):
        f"{self.x}" 
    
    
obj2 = Sum()
print(obj2.num(10,20)) 
print(obj2.add())       
```

```python
class Number: #parent class
    def Num(self,x,y):
        self.x = x
        self.y = y
        return f"x :{x} and y:{y} "  
class Addition(Number):
    def sum(self):
        return f"{self.x + self.y}"  

obj1 = Addition() #creating obj of child class
print(obj1.Num(10,20))# to perform add we req value of x and y and this values are from fun num so called it
print(obj1.sum() )  # sum fun call using child class obj 
          
```

```python
class Parent:
    def P_Name(self,Name):
        self.Name = Name
        
class Child(Parent):
    def C_Name(self,Name1):
        self.Name1 = Name1 
        return f"{Name1}  {self.Name}"
    
My_Name = Child()
My_Name.P_Name("Vitthal Dhage ")
My_Name.C_Name("Darshan")       
            
```

```python
class Bank:
    def User_Name(self,Name,Add,Mob):
        self.Name = Name
        self.Add = Add
        self.Mob = Mob
        return f"Details of Customer \nName : {Name} \nAdd : {Add} and \nMobile number : {Mob}"
class Credit_card(Bank):
    def Card(self,Income):
        self.Income = Income 
        return f"Credit card details \nName : {self.Name} \nAdd : {self.Add} \nMob : {self.Mob} and \nIncome :{Income}"  
    
C1 = Credit_card()
print(C1.User_Name("Tushar","Pune","1234567890"))
print(C1.Card("10LPA"))      
        
```
