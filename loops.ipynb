```python
#Wap to first half in asc and second half in dec.
#l1 = [1,3,2,4,5,6]
#l1 = [1,2,3,6,5,4]
l1 = [1,3,2,4,5,6]
mid = len(l1)//2 # 6//2 == 3
first = sorted(l1[:mid]) # 0:3...1,2,3
second = sorted(l1[mid:],reverse = True)# 6,5,4
print(first+second) # 1,2,3,6,5,4
```
