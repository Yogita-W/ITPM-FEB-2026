```python
#Single Inheritance
class Cal :
    def Number(self,x,y):
        self.x = x
        self.y = y
        return f"X : {x} and Y : {y}"
class Addition(Cal):
    def add(self):
        return f"Addition of x and y : {self.x + self.y}"

obj1 = Addition()
print(obj1.Number(10,20))
print(obj1.add())
```

```python
class Parent1 :
    def Father(self,Name):
        self.Name = Name
class Parent2:
    def Mother(self,Name1):
        self.Name1 = Name1

class child(Parent1,Parent2):
    def child(self,Name2):
        self.Name2 = Name2
        return f" My name is {Name2} {self.Name1} {self.Name}"

obj1 = child()
obj1.Father("Ramesh Ugale")
obj1.Mother("Savita")
obj1.child("Kartik")
```

```python
#Multiple Inheritance
class Cal1:
    def Number1(self,x,y):
        self.x = x
        self.y = y
        return f"X : {x} and Y : {y}"
class Cal2:   
    def Number2(self,a,b):
        self.a = a
        self.b = b
        return f"A : {a} and B : {b}"
class Addition(Cal1,Cal2):
    def add(self):
        return f"Addition of x ,y and a ,b: {self.x + self.y+self.a + self.b}"

obj1 = Addition()
print(obj1.Number1(10,20))
print(obj1.Number2(30,40))
print(obj1.add())
```

```python
#Multi-level
class GrandParent():
    def GP(self,Hname):
        self.Hname = Hname
        
class Parent(GrandParent):
    def P(self,Hname1):
        self.Hname1 = Hname1
        return f"My Home names are {Hname1} and {self.Hname}"

class Child(Parent):
    def C(self,Hname2):
        self.Hname2 = Hname2
        return f"My Home Names are {Hname2} { self.Hname1} and {self.Hname}"

obj1 = Child()
obj1.GP("40Gav")
print(obj1.P("Jalgaon"))
print(obj1.C("Pune"))
        
```

```python
class A : 
    def a(self,a):
        self.a = a
        return f"a:{a}"

class B(A):
    def b(self,b):
        self.b = b
        return f"a:{self.a} and b: {self.b}"

obj1 = B()
print(obj1.a(10))
print(obj1.b(20))
```

```python
class Cal:
    def Number(self,a,b):
        self.a = a
        self.b = b
        return f"A:{a} and B:{b}"

class Addition(Cal):
    def Add(self):
        return f"Addition : {self.a + self.b}"

class Sub(Cal):
    def S(self):
        return f"Substraction : {self.a - self.b}"

class Mul(Cal):
    def M(self):
        return f"Multiplication : {self.a * self.b}"

obj1 = Addition()
print(obj1.Number(10,20))
print(obj1.Add())

obj2 = Sub()
print(obj2.Number(30,20))
print(obj2.S())

obj3 = Mul()
print(obj3.Number(5,10))
print(obj3.M())

```

```python
class Trainer :
    def Domain(self,Dname):
        self.Dname = Dname

class Std(Trainer):
    def Sname1(self,name1):
        self.name1 = name1
        return f" I am {name1} and AM from {self.Dname} Domain"

class Std1(Trainer):
    def Sname2(self,name2):
        self.name2 = name2
        return f" I am {name2} and AM also from {self.Dname} Domain "

class Company(Std,Std1):
    def placed(self):
        return f"My name is {self.name2} am placed"

obj1 = Company()
obj1.Domain("Ds+DA")
obj1.Sname1("Darshan")
obj1.Sname2("Krushna")
obj1.placed()

    
```
