```python
def Anagram(str1,str2):
    str1 = input("Enter String")
    str2 = input("enter string")
    l1 = list(str1)
    l1.sort()
    l2 = list(str2)
    l2.sort()
    print(l1,l2)
    if l1 == l2:
        print("Strings are Anagram")
    else :
        print("Strings are Not Anagram") 

Anagram(str1,str2)        
```

```python
# to check if given no is prime or not
def prime(n): #Create function
  count = 0   # c=0,c=1,c=2
  for i in range(1,n+1):#(1,6) i = 1,i=2
      if n % i == 0:# 5%1==0,5%2!=0,5%3!=0,4!=0,5%5==0
        count +=1 # c=1,2
  if count==2:
    print("prime")
  else:
    print("not a prime")
r=int(input())
prime(r)
```

```python
#wap to print prime numbers from list
list1 = [22,35,21,43,47,66,64]
prime_number = list(filter(lambda x : x>1 and all (x%i!=0 for i in range(2,x)),list1))
#                                                  for i in range(2,x):
#                                                       x%i!=0
#                                                                 (2,43).. 43%2,43%3,43%4......43%42
print(prime_number)
```

```python
#WAp to sort given list in ascending order ueing lambda function.
l1 = ["Python","Apti","DA","Softskill","UI"]
#o/p ["Apti","DA","Python","softskill","UI"]
#l2 = sorted(l1,key = lambda l2 : l2)
l2 = sorted(l1,key = lambda x : x)
l3 = sorted(l1,key = lambda x : x ,reverse=True)
print(l2)
print(l3)
```

```python
#WAp to sort given list in ascending order using size of word using lambda function.
l1 = ["Python","Apti","DA","Softskill","UI"]
#o/p ["DA","UI","Apti","Python","softskill"]
l2 = list(sorted(l1, key = lambda x : len(x)))
print(l2)
```
